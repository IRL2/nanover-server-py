# Configuring visuals for iMD-XR client

This notebook demonstrates how to use selections and rendering configuration to show a protein in a cartoon visualisation, and a ligand in liquorice visualisation.

## Simulation setup

First we load a protein with ligand system in the usual way:

```python
from nanover.openmm import OpenMMSimulation

simulation = OpenMMSimulation.from_bundle_path("../example_systems/trypsin_benzamidine.openmm.zip")
simulation.load()
```

## Selection extraction

Next we grab a frame from the simulation and convert it to an MDAnalysis universe, which we use to query the topology for the atoms of interest:

```python
from nanover.mdanalysis import frame_data_to_mdanalysis

universe = frame_data_to_mdanalysis(simulation.make_topology_frame())

structure_atoms = universe.select_atoms("not resname BEN")
molecule_atoms = universe.select_atoms("resname BEN")
```

## Server setup

Then we setup NanoVer to run the simulation and serve it over the network in the usual way:

```python
from nanover.app import OmniRunner

imd_runner = OmniRunner.with_basic_server(simulation, port=0, name="visual configuration example")
imd_runner.load(0)
```

## Visual configuration

Finally, we use the utilities to make two selections using the atoms from earlier, specifying which named renderers they should use:

```python
from nanover.jupyter import NanoverJupyterUtilities

utilities = NanoverJupyterUtilities.from_runner(imd_runner)

utilities.selections.update_selection("protein", renderer="cartoon", particle_ids=structure_atoms.indices)
utilities.selections.update_selection("ligand", renderer="liquorice", particle_ids=molecule_atoms.indices)
```

## Viewing in virtual reality

Finally, we connect using the [NanoVer iMD-XR client](https://irl2.github.io/nanover-docs/installation.html#installing-the-imd-xr-client) to see and interact with the live OpenMM dynamics.
